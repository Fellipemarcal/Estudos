```python
import pandas as pd

df = pd.DataFrame({
    "produit": ["A","A","A","B","B","B"],
    "quantite_vendue": [100,10,15,30,28,32]
})
#nous pouvons utiliser ça directement dans mon DataFrame?
result = df.groupby("produit")["quantite_vendue"].agg(["sum","mean","std","max","min"])

print(result)
```
