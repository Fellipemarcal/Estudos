```python
import pandas as pd
import matplotlib.pyplot as plt
df1 = pd.DataFrame({
    "produit": ["A","A","A","B","B","B"],
    "quantite_vendue": [100,10,15,30,28,32]
})

df2 = pd.DataFrame({
    "produit": ["A","A","A","B","B","B"],
    "quantite_vendue": [140,15,25,330,18,52]
})

result = pd.concat({
    'jan':df1,
    'fev':df2
},
   names=['mois']               
                  )


df1["mois"] = "jan"
df2["mois"] = "fev"

df_total = pd.concat([df1, df2])
```

```python
a = result.groupby('produit')['quantite_vendue'].agg(['sum','mean','max','min'])
```

```python
ax = a.plot(kind='bar',figsize = (15,5))

for i in ax.containers:
    ax.bar_label(i)
plt.show()
```
